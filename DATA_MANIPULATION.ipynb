```python
import numpy as np

prices = np.random.randint(100, 5001, 20)

print("Original Prices:", prices)
print("First 5:", prices[:5])
print("Last 5:", prices[-5:])

gst_prices = prices * 1.18
print("Prices with 18% GST:", gst_prices)

print("Above 2500:", prices[prices > 2500])
print("Between 1000 and 3000:", prices[(prices >= 1000) & (prices <= 3000)])

print("Minimum:", np.min(prices))
print("Maximum:", np.max(prices))
print("Mean:", np.mean(prices))
print("Median:", np.median(prices))
print("Variance:", np.var(prices))
print("Standard Deviation:", np.std(prices))

print("Total Value:", np.sum(prices))
print("Products above average:", np.sum(prices > np.mean(prices)))

np.savetxt("products.csv", gst_prices, delimiter=",")
print("products.csv saved successfully")
```

```python
import numpy as np

data = np.random.randint(1, 1001, 50)

print("Original Data:", data)
print("First 10:", data[:10])
print("Last 10:", data[-10:])
print("Every Second Element:", data[::2])
print("Indexes 5, 10, 15, 20:", data[[5, 10, 15, 20]])

data = data + 100
print("After Adding 100:", data)

data = data * 2
print("After Multiplying by 2:", data)

print("Above 500:", data[data > 500])
print("Between 200 and 600:", data[(data >= 200) & (data <= 600)])
print("Even Numbers:", data[data % 2 == 0])
print("Odd Numbers:", data[data % 2 != 0])

print("Sum:", np.sum(data))
print("Product:", np.prod(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))

print("25th Percentile:", np.percentile(data, 25))
print("50th Percentile:", np.percentile(data, 50))
print("75th Percentile:", np.percentile(data, 75))

print("Numbers above mean:", np.sum(data > np.mean(data)))

print("Sorted Data:", np.sort(data))

np.save("original_data.npy", data)
np.savetxt("final_data.csv", data, delimiter=",")

print("original_data.npy saved successfully")
print("final_data.csv saved successfully")
```
